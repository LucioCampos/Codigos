```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
imagem = cv2.imread('navioescola.png') # lendo imagem em RGB
x,y,z=imagem.shape #encontrando tamanho da imagem
zeros=np.zeros((x,y), dtype = "uint8") #matriz de zeros do tamanho da imagem
(vermelho, verde, azul) = cv2.split(imagem) #separando os canais R, G e B
# Gerando as imagens de apenas um canal por vez
imvermelho=cv2.merge([vermelho, zeros, zeros]) #canal vermelho
imverde=cv2.merge([zeros, verde, zeros]) #canal verde
imazul=cv2.merge([zeros, zeros, azul]) #canal azul
plt.figure(figsize=(15,10))
plt.subplot (1,3,1)
plt.imshow(imvermelho)
plt.title('Canal vermelho'),plt.xticks([]), plt.yticks([])
plt.subplot (1,3,2)
plt.imshow(imverde)
plt.title('Canal verde'),plt.xticks([]), plt.yticks([])
plt.subplot (1,3,3)
plt.imshow(imazul)
plt.title('Canal azul'),plt.xticks([]), plt.yticks([])
plt.show()
```
