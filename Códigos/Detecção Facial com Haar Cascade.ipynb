```python
import cv2
from matplotlib import pyplot as plt
imagem = cv2.imread('varios_.jpeg') # carregando a imagem teste
imagem  = cv2.cvtColor(imagem, cv2.COLOR_RGB2BGR)
classificador_face = cv2.CascadeClassifier(r"/content/haarcascade_frontalface_default.xml")
gray = cv2.cvtColor(imagem, cv2.COLOR_BGR2GRAY)
face_detector = classificador_face.detectMultiScale(gray, scaleFactor=1.1, minSize=(4,4))
#print("Faces encontradas: ", len(face_detector))
for (x, y, l, a) in face_detector:
    cv2.rectangle(imagem, (x, y), (x + l, y + a), (0, 255, 0), 4)
cv2.putText(imagem,f'Faces Detectadas: {str(len(face_detector))}', (650, 50), cv2.FONT_HERSHEY_SIMPLEX, 1.2, (255, 0, 0), 4)
plt.figure(figsize=(12,8))
plt.imshow(imagem),plt.xticks([]), plt.yticks([])
plt.show()
```
