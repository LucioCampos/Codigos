```python
import numpy as np
import cv2
from matplotlib import pyplot as plt
img = cv2.imread('lenna.jpg')
kernel1=cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3,3))
kernel2=cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5,5))
gradient1 = cv2.morphologyEx(img, cv2.MORPH_GRADIENT, kernel1)
gradient2 = cv2.morphologyEx(img, cv2.MORPH_GRADIENT, kernel2)
plt.figure(figsize=(10,7))
plt.subplot(1,3,1)
plt.imshow(img, cmap = 'gray')
plt.title('Original'), plt.xticks([]), plt.yticks([])
plt.subplot(1,3,2)
plt.imshow(gradient1, cmap = 'gray')
plt.title('Gradiente Morfológico 3x3'), plt.xticks([]), plt.yticks([])
plt.subplot(1,3,3)
plt.imshow(gradient2, cmap = 'gray')
plt.title('Gradiente Morfológico 5x5'), plt.xticks([]), plt.yticks([])
plt.show()
```
