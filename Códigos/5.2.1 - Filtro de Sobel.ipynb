```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
img = cv2.imread('/content/drive/MyDrive/Livro visão/Códigos/pripyat2.jpg',0)
# SOBEL=(img, var_arm, h, v, ksize)
#img é imagem de entrada,
#var_arm é a codificação da imagem
#h ou v a direção do filtro
#ksize o tamanho nxn do kernel.
sobelx = cv2.Sobel(img, cv2.CV_8U, 0, 1, ksize=3) # kernel 3x3
sobely = cv2.Sobel(img, cv2.CV_8U, 1, 0, ksize=3) # kernel 3x3
plt.figure(figsize=(20,15))
plt.subplot (1,3,1)
plt.imshow(img,cmap=plt.cm.gray)
plt.title('Original'),plt.xticks([]), plt.yticks([])
plt.subplot (1,3,2)
plt.imshow(sobelx,cmap=plt.cm.gray)
plt.title('Sobel na direção x'),plt.xticks([]), plt.yticks([])
plt.subplot (1,3,3)
plt.imshow(sobely,cmap=plt.cm.gray)
plt.title('Sobel na direção y'),plt.xticks([]), plt.yticks([])
```
