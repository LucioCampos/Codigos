```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
img = cv2.imread('/content/drive/MyDrive/Livro visão/Códigos/pripyat2.jpg',0)
#LAP=Laplacian(img, var_arm)
#img: imagem de entrada
#var_arm: codificação da imagem
lap = cv2.Laplacian(img, cv2.CV_8U)
plt.figure(figsize=(25,20))
plt.subplot (1,2,1)
plt.imshow(img,cmap=plt.cm.gray)
plt.title('Original'),plt.xticks([]), plt.yticks([])
plt.subplot (1,2,2)
plt.imshow(lap,cmap=plt.cm.gray)
plt.title('Laplaciano'),plt.xticks([]), plt.yticks([])
```
