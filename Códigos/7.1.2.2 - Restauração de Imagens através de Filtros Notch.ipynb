```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
from math import sqrt,exp
img = cv2.imread("indiano.png",0)
fourier = np.fft.fft2(img)
center = np.fft.fftshift(fourier)
centerlog =np.log(1+np.abs(center))
plt.figure(figsize=(8,5))
plt.subplot (1,2,1)
plt.imshow(img,cmap=plt.cm.gray),plt.xticks([]), plt.yticks([])
plt.title('Original'),plt.xticks([]), plt.yticks([])
plt.subplot (1,2,2)
plt.imshow(centerlog,cmap=plt.cm.gray)
plt.title('Espectro Original'),plt.xticks([]), plt.yticks([])
plt.show()
```

```python
#Criando o filtro através da visualização dos ruídos no espectro de Fourier
rows, cols = centerlog.shape
crow,ccol = rows//2 , cols//2
mask = np.ones((rows,cols),np.uint8)
#desenhando os círculos no espectro
preto=(0,0,0) #definindo cor
circ1=cv2.circle(mask, (120, 16), 16, preto,-1)
circ2=cv2.circle(mask, (120, 42), 16, preto,-1)
circ3=cv2.circle(mask, (120, 101), 16, preto,-1)
circ4=cv2.circle(mask, (120, 218), 16, preto,-1)
circ5=cv2.circle(mask, (120, 278), 16, preto,-1)
circ6=cv2.circle(mask, (120, 301), 16, preto,-1)
plt.figure(figsize=(8,5))
plt.imshow(mask,cmap=plt.cm.gray),plt.xticks([]), plt.yticks([])
plt.show()
```

```python
fshift=mask*center
f_ishift = np.fft.ifftshift(fshift)
inverse = np.fft.ifft2(f_ishift)
plt.figure(figsize=(8,5))
plt.subplot (1,2,1)
plt.imshow(np.log(1+np.abs(fshift)),cmap=plt.cm.gray),plt.xticks([]), plt.yticks([])
plt.title('Espectro Filtrado'),plt.xticks([]), plt.yticks([])
plt.subplot (1,2,2)
plt.imshow(np.abs(inverse),cmap=plt.cm.gray),plt.xticks([]), plt.yticks([])
plt.title('Imagem Filtrada'),plt.xticks([]), plt.yticks([])
plt.show()
```
