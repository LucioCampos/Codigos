```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
imagem = cv2.imread('flores.JPG')
imgBGR = cv2.cvtColor(imagem, cv2.COLOR_RGB2BGR)
x,y,z=imagem.shape #encontrando tamanho da imagem
zeros=np.zeros((x,y), dtype = "uint8")
(r,g,b) = cv2.split(imagem) #separando os canais RGB
red = cv2.equalizeHist(r)
green = cv2.equalizeHist(g)
blue = cv2.equalizeHist(b)
equalized=cv2.merge([blue, green, red])
plt.figure(figsize=(15,10))
plt.subplot (1,2,1)
plt.imshow(imgBGR)
plt.title('Imagem Original'),plt.xticks([]), plt.yticks([])
plt.subplot (1,2,2)
plt.imshow(equalized)
plt.title('Imagem equalizada'),plt.xticks([]), plt.yticks([])
plt.show()
```
