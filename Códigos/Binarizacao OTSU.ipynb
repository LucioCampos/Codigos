# Binarização por OTSU

```python
# Carregando Bibliotecas
import cv2
import numpy as np
from matplotlib import pyplot as plt
%notebook inline
```

```python
#Carregando a imagem e as funções
img=cv2.imread('cafe.jpg',0)
metodo = cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU
ret, imgBin = cv2.threshold(img, 0, 255, metodo)
print("Treshold:", ret)
```

```python
print ('O melhor Treshold é')
plt.hist(img.ravel(), 256, (0,256))
cv2.imshow('Original', img)
plt.show()
cv2.imshow('Otsu', imgBin)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
