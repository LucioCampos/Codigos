```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
img = cv2.imread('olho.PNG',0)
imgGauss = cv2.medianBlur(img, 7) #suavização para melhorar o resultado
imgBin = cv2.adaptiveThreshold(imgGauss, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 9, 3)
plt.figure(figsize=(15,10))
plt.subplot (2,2,1)
plt.imshow(img,cmap=plt.cm.gray), plt.xticks([]), plt.yticks([])
plt.title('Original')
plt.subplot (2,2,2)
plt.hist(img.ravel(), 256, [0,256])
plt.title('Histograma')
plt.figure(figsize=(15,10))
plt.subplot (2,2,1)
plt.imshow(imgBin,cmap=plt.cm.gray), plt.xticks([]), plt.yticks([])
plt.title('Binarização por OTSU')
plt.subplot (2,2,2)
plt.hist(imgBin.ravel(), 256, [0,256])
plt.title('Histograma OTSU')
plt.show
```
