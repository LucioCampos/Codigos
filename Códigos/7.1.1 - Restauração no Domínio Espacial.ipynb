```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
img=cv2.imread("brain.jpg",0)
imgr1 = cv2.medianBlur(img, 3) #Filtragem da mediana, kernel 3x3
imgr2 = cv2.blur(img, (3,3)) #Filtragem da media, kernel  5x5
imgr3 = cv2.GaussianBlur(img, (3,3), 2) #Filtragem gaussiana,kernel 5x5,sigma=2
#Mostrando os resultados
plt.figure(figsize=(15,10))
plt.subplot (1,4,1)
plt.imshow(img,cmap=plt.cm.gray)
plt.title('Original'),plt.xticks([]), plt.yticks([])
plt.subplot (1,4,2)
plt.imshow(imgr1,cmap=plt.cm.gray)
plt.title('Filtro da Mediana'),plt.xticks([]), plt.yticks([])
plt.subplot (1,4,3)
plt.imshow(imgr2,cmap=plt.cm.gray)
plt.title('Filtro da Média'),plt.xticks([]), plt.yticks([])
plt.subplot (1,4,4)
plt.imshow(imgr3,cmap=plt.cm.gray)
plt.title('Filtro Gaussiano'),plt.xticks([]), plt.yticks([])
plt.show()
```
