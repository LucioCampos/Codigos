```python
#Importando bibliotecas
import cv2
import imutils
import dlib
from imutils import face_utils
from matplotlib import pyplot as plt
```

```python
imagem = cv2.imread('varios_.jpeg') # carregando a imagem teste
imagem  = cv2.cvtColor(imagem, cv2.COLOR_RGB2BGR)
#path = "shape_predictor_68_face_landmarks.dat"
detector  = dlib.get_frontal_face_detector()
faces_detectadas = detector(imagem,1)
#print(faces_detectadas)
#print("Faces Detectadas: ", len(faces_detectadas))

for face in faces_detectadas:

    e, t, d, b = (int(face.left()), int(face.top()), int(face.right()), int(face.bottom()))
    cv2.rectangle(imagem, (e, t), (d, b), (0, 255, 0), 2)
cv2.putText(imagem,f'Faces Detectadas: {str(len(faces_detectadas))}', (650, 50), cv2.FONT_HERSHEY_SIMPLEX, 1.2, (255, 0, 0), 4)
plt.figure(figsize=(12,8))
plt.imshow(imagem),plt.xticks([]), plt.yticks([])
plt.show()
```
