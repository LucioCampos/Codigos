```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
img = cv2.imread('lowcontrast1.jpg',0)
h_eq = cv2.equalizeHist(img)
imgEqualizada = cv2.equalizeHist(img)
plt.figure(figsize=(8,4))
plt.subplot (2,2,1)
plt.imshow(img,cmap=plt.cm.gray)
plt.title('Imagem Original'),plt.xticks([]), plt.yticks([])
plt.subplot (2,2,2)
plt.title("Histograma Original"),plt.xticks([]), plt.yticks([])
plt.xlabel("Intensidade")
plt.ylabel("Número de Pixels")
plt.hist(img.ravel(), 256)
plt.figure(figsize=(8,4))
plt.subplot (2,2,1)
plt.imshow(imgEqualizada,cmap=plt.cm.gray)
plt.title('Imagem Equalizada'),plt.xticks([]), plt.yticks([])
plt.subplot (2,2,2)
plt.title("Histograma Equalizado"),plt.xticks([]), plt.yticks([])
plt.xlabel("Intensidade")
plt.ylabel("Número de Pixels")
plt.hist(imgEqualizada.ravel(), 256)
plt.show ()
```
