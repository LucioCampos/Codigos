```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
img=cv2.imread("/content/drive/MyDrive/Livro visão/Códigos/median_test.jpg",0)
imgr1 = cv2.medianBlur(img, 3) # Filtragem com janela de  3x3
imgr2 = cv2.medianBlur(img, 5) # Filtragem com janela de  5x5
imgr3 = cv2.medianBlur(img, 9) # Filtragem com janela de  9x9
plt.figure(figsize=(15,10))
plt.subplot (1,4,1)
plt.imshow(img,cmap=plt.cm.gray)
plt.title('Original'),plt.xticks([]), plt.yticks([])
plt.subplot (1,4,2)
plt.imshow(imgr1,cmap=plt.cm.gray)
plt.title('Filtragem 3x3'),plt.xticks([]), plt.yticks([])
plt.subplot (1,4,3)
plt.imshow(imgr2,cmap=plt.cm.gray)
plt.title('Filtragem 5x5'),plt.xticks([]), plt.yticks([])
plt.subplot (1,4,4)
plt.imshow(imgr3,cmap=plt.cm.gray)
plt.title('Filtragem 9x9'),plt.xticks([]), plt.yticks([])
```
