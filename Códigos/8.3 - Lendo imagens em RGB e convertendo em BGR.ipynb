```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
imgRGB = cv2.imread('navioescola.png')
imgBGR = cv2.cvtColor(imgRGB, cv2.COLOR_RGB2BGR)
plt.figure(figsize=(10,6))
plt.subplot (1,2,1),plt.xticks([]), plt.yticks([])
plt.imshow(imgRGB,cmap=plt.cm.gray)
plt.title('Imagem RGB')
plt.subplot (1,2,2)
plt.imshow(imgBGR,cmap=plt.cm.gray)
plt.title('Imagem BGR'),plt.xticks([]), plt.yticks([])
plt.show()
```
