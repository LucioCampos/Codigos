```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
img = cv2.imread('/content/drive/MyDrive/Livro visão/Códigos/pripyat2.jpg',0)
#edges = cv2.Canny(img,MinVal,MaxVal)
# MinVal e MaxVal: valores mínimos e máximos da limiarização
#por histerese aplicada no ultimo passo do algoritmo
canny = cv2.Canny(img,70,150)
plt.figure(figsize=(25,20))
plt.subplot (1,2,1)
plt.imshow(img,cmap=plt.cm.gray)
plt.title('Original'),plt.xticks([]), plt.yticks([])
plt.subplot (1,2,2)
plt.imshow(canny,cmap=plt.cm.gray)
plt.title('Filtragem por Canny'),plt.xticks([]), plt.yticks([])
```
